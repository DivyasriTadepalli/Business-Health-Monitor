```python
import pandas as pd
data=pd.read_excel("C:/Users/divya/OneDrive/Documents/BrandX_India_Dataset_Analysis.xlsx")
# data.head()
data.shape
# data.columns
# data.info()
```

```python
promotion_analysis = data.groupby("Promotion_Type").agg(Records=("Revenue", "count"),Avg_Revenue=("Revenue", "mean"),
                                                        Avg_Profit_Margin=("Profit_Margin", "mean"),Avg_Discount=("Max_Discount_Offered", "mean"),
                                                        Avg_Promotion_Impact=("Promotion_Impact", "mean")).round(2)
print(promotion_analysis)
```

```python
import matplotlib.pyplot as plt
promotion_analysis = data.groupby("Promotion_Type")["Profit_Margin"].mean().round(2)
# print(promotion_analysis)
plt.bar(promotion_analysis.index,promotion_analysis.values,color="#8FB9D8",edgecolor="#17365D")
plt.xlabel("Promotion Type",color="#17365D")
plt.ylabel("Average Profit Margin",color="#17365D")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.2)
plt.title("Average Profit Margin by Promotion Type",color="#17365D",fontsize=14,fontweight="bold")
plt.show()
```
